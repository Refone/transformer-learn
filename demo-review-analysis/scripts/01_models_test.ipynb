````python
print(model)
````

````python
# 还可以指定分类数量
model4sc2 = AutoModelForSequenceClassification.from_pretrained(path, num_labels=5)
````

# 前向传播

````python
import torch

vocab_size = 1000
N = 4
L = 6

input_ids = torch.randint(low=0, high=vocab_size, size=(N, L))
print(input_ids.shape, input_ids.dtype)
print(input_ids)
````

## 原始模型的前向传播

````python
output = model(input_ids)
print(output)
````

````python
print(output.last_hidden_state.shape)
print(output.pooler_output.shape)
````

## 带任务头的模型

````python
output = model4sc2(input_ids)   # 5 分类模型，输入 (N, L)，输出 (N, 5)
print(output.logits.shape)  # (N, 5)
print(output)
````

````python
# 定义目标值
targets = torch.randint(low=0, high=5, size=(N,))
print(targets)
````

````python
output = model4sc2(input_ids, labels=targets)
output.loss.backward()  # 反向传播计算梯度
print(output.logits)  # (N, 5)
print(output.loss)  # 输出损失值
````
